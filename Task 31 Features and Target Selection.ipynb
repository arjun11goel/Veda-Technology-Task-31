# Task 31: Features and Target Selection

## Objective
Understand the fundamental structure of supervised machine learning problems by distinguishing between input features ($X$) and the target variable ($y$) using the Iris Dataset.

## Overview
In supervised learning, algorithms learn a mapping function from input variables (features) to an output variable (target). 
- **Features ($X$):** The measurable properties or characteristics used as inputs by the model.
- **Target ($y$):** The label or outcome variable that the model is trained to predict.

```python
import pandas as pd

# 1. Load the Iris Dataset
df = pd.read_csv('iris_dataset.csv')

# Display the first 5 rows to understand the structure
print("--- First 5 Rows of the Iris Dataset ---")
print(df.head(), "\n")

# 2. Define Features (X) and Target (y)
# Features: The measurable flower dimensions used as inputs
feature_columns = ['sepal length (cm)', 'sepal width (cm)', 'petal length (cm)', 'petal width (cm)']
X = df[feature_columns]

# Target: The species column that we want the model to predict
target_column = 'species'
y = df[target_column]

# 3. Output the selection details
print("--- Selected Feature Columns (X) ---")
print(X.head(3), "\n")

print("--- Selected Target Column (y) ---")
print(y.head(3), "\n")

print(f"Shape of Features (X): {X.shape}")
print(f"Shape of Target (y): {y.shape}")
```

## Selection Reasoning

1. **Selected Feature Columns ($X$):** 
   - `sepal length (cm)`, `sepal width (cm)`, `petal length (cm)`, `petal width (cm)`
   - **Reasoning:** These numerical measurements capture the physical dimensions of the iris flowers. They provide objective, quantitative input data that machine learning algorithms can analyze to find patterns distinguishing one species from another.

2. **Selected Target Column ($y$):** 
   - `species`
   - **Reasoning:** This is the categorical label indicating the specific iris variety (e.g., *setosa*, *versicolor*, *virginica*). Since the ultimate goal of our classification model is to predict the correct flower category given its measurements, `species` serves as the target variable.

3. **Avoiding Data Leakage:**
   - We strictly separate $X$ and $y$ and ensure the target column is excluded from the feature set to prevent data leakage, ensuring the model learns generalizable patterns rather than cheating by inspecting the answer.

# Interview Q&A: Supervised Machine Learning Fundamentals

### Q1: What is a feature?
* **Answer:** A feature is an individual measurable property, characteristic, or input variable used by a machine learning model to make predictions. In tabular datasets, features correspond to columns (other than the target label). For example, in the Iris dataset, the sepal and petal dimensions are features.

### Q2: What is a target variable?
* **Answer:** The target variable (also known as the label, dependent variable, or response variable) is the specific outcome or value that the machine learning model is designed to predict. In classification tasks, it represents the category/class; in regression tasks, it represents a continuous numerical value.

### Q3: What is the difference between $X$ and $y$ in machine learning?
* **Answer:** 
  - **$X$ (Capital X):** Represents the matrix of input features. It typically has a 2D shape of `(n_samples, n_features)` containing all the independent variables.
  - **$y$ (Lowercase y):** Represents the vector or array of target values. It has a 1D shape of `(n_samples,)` containing the corresponding ground-truth labels or outcomes that the model learns to map from $X$.
